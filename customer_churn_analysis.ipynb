# Final Business Insights

1. The majority of customers remained with the company, but a significant number churned, highlighting the importance of customer retention.

2. Customers with month-to-month contracts had the highest churn rate, whereas customers with one-year and two-year contracts were much more likely to stay.

3. Customers with higher monthly charges showed a greater tendency to churn compared to customers with lower monthly charges.

4. Customers with shorter tenure were more likely to leave the company, indicating that retaining new customers is a key challenge.

5. Customers without partners or dependents experienced higher churn than those with partners or dependents.

6. Contract type, monthly charges, and customer tenure were the strongest factors associated with customer churn.

7. The company can reduce churn by encouraging long-term contracts, improving customer support for new customers, and offering personalized retention strategies.
